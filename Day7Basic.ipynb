```python
def check_for_word():

    Word= str(input("your text"))
    data= True
    line_no=1
    with open("practice.txt","r") as f:
        while data:
            data=f.readline()
            if(Word in data):
                print(line_no)
                return
            line_no+=1
    return -1
check_for_word()
```

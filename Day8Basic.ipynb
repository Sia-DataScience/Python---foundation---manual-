```python
class Student:
    collage_name="ABC college"#class attribute
    name="not mentioned"
     # paramerized constructor
    def __init__(self,name,marks):
        self.name= name #object attribute
        self.marks= marks

    def welcome(self):
        print("welcome student,", self.name)
    def mark_get(self):
        print("your marks,",self.marks)
    @staticmethod # static method (decorator)
    def hello():
        print("hello")

S1=Student("shrawani",97)
print(S1.welcome())
print(S1.mark_get())
S1.hello()
S2=Student("Amruta",95)
print("Your name:",S2.name,"your marks:",S2.marks)

S3=Student("Emily",98)
print(S3.name,S3.marks)

print(S1.collage_name)
```

```python
# practice questions
class Student:
    def __init__(self, name, marks ):
        self.name= name
        self.marks=marks
    def get_avg(self):
        sum=0
        for val in self.marks:
            sum+=val
        print("hi:",self.name,"your avg score is:", sum/3)
S1=Student("tomiyoka giyu",[99,98,97])
S1.get_avg()
```

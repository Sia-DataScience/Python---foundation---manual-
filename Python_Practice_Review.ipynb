```python
#practice question 1:-
# The Dynamic calculator function:-
def calculator(a,b, operation):
    if operation=="add":
        return a+b
    elif operation=="substraction":
        return a-b
    elif operation=="multiply":
        return a*b
    elif operation=="divide":
        return a/b
Numbers= calculator(10,2,"add")
print(Numbers)
Numbers= calculator(10,2,"substraction")
print(Numbers)
Numbers= calculator(10,2,"multiply")
print(Numbers)
Numbers= calculator(10,2,"divide")
print(Numbers)
Numbers= calculator(100,2,"divide")
print(Numbers)
```

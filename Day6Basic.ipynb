```python
#2)calculating a factorial
def cal_fact(n):
    fact=1
    for i in range(1,n+1):
        fact *= i
    print(fact)
cal_fact(6)
```

```python
# recursive function
def show(n):
    if(n == 0):
       return
    print(n)
    show(n-1)
    print("end")
show(3)
```

```python
def fact(n):
    if(n==1 or n==0):
        return 1
    else:
        return fact(n-1)*n
    print(fact)
fact(6)
```

```python
# practice questions 2
def print_list(list,idx=0):
    if (idx==len(list)):
        return
    print(list[idx])
    print_list(list, idx+1)
Fruit=["apple","mango","banana","watermelon","litchi"]
print_list(Fruit)
```
